```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point
from shapely import wkt
```

```python
# Load the bus route data
bus_data = pd.read_csv('stop ids with time diff and location2.csv')
# Check for NaN values in the location columns
bus_data = bus_data.dropna(subset=['stop_lat_1', 'stop_lon_1'])
```

```python
bus_data
```

```python
# Create a GeoDataFrame for stops
geometry = [Point(xy) for xy in zip(bus_data['stop_lat_1'], bus_data['stop_lon_1'])]
stops_gdf = gpd.GeoDataFrame(bus_data, geometry=geometry)
```

```python
censustract=pd.read_csv('Census_2020__Tracts_for_San_Francisco_20250310.csv')
```

```python
censustract
```

```python
# Assuming the census tract data includes geometry information in WKT format
censustract['geometry'] = censustract['the_geom'].apply(wkt.loads)
tracts_gdf = gpd.GeoDataFrame(censustract, geometry='geometry')
```

```python
tracts_gdf
```

```python

# Ensure both GeoDataFrames use the same coordinate reference system (CRS)
stops_gdf = stops_gdf.set_crs('EPSG:4326')
tracts_gdf = tracts_gdf.set_crs('EPSG:4326')
```

```python
# Perform a spatial join
mapped_data = gpd.sjoin(stops_gdf, tracts_gdf, how='left', predicate='within')
```

### Where are high ranked and low-ranked schools located? 
In this (hopefully) short notebook, we explore the locations of the top schools in San Francisco. Are they equitably located across the city? School rankings are based on an index score based on standardized test scores created by our teammate Ronald. 

```python
#libraries
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import contextily as ctx
import numpy as np
import branca.colormap as cm
import plotly.express as px
import folium
import branca.element #just in case for adding titles
```

```python
# school location data
schools = gpd.read_file('data/Filtered_Schools.csv')
schools = schools[schools['CCSF Entity'].str.contains('SFUSD')]
schools = schools[schools['Grade Range'].str.contains('9-12')]
schools = schools[['Campus Name', 'Campus Address', 'Location 1']]
schools = schools.drop(index=38)

schools[['latitude', 'longitude']] = schools['Location 1'].str.extract(r'\(([^,]+), ([^,]+)\)').astype(float)
schools.head(18)
```

```python
#top 5 high schools 
m = folium.Map(location=[37.7619, -122.4194], 
                   zoom_start=12,
                   tiles='CartoDB positron', 
                   attribution='CartoDB')

#George Washington High School
folium.Marker(
    location=[37.777905, -122.491013],
    popup='Washington, George Washington High School',
    icon=folium.Icon(color='blue', icon='icon')
).add_to(m)

#Lincoln HS
folium.Marker(
    location=[37.746594, -122.480240],
    popup='Lincoln, Abraham Lincoln High School',
    icon=folium.Icon(color='blue', icon='icon')
).add_to(m)

#Gateway HS
folium.Marker(
    location=[37.783264, -122.436691],
    popup='Gateway High School / Kipp Sf Bay Academy',
    icon=folium.Icon(color='blue', icon='icon')
).add_to(m)

# Galileo HS
folium.Marker(
    location=[37.803791, -122.424149],
    popup='Galileo High School',
    icon=folium.Icon(color='blue', icon='icon')
).add_to(m)

# Burton HS
folium.Marker(
    location=[37.721546, -122.406555],
    popup='Burton, Phillip And Sala Burton High School',
    icon=folium.Icon(color='blue', icon='icon')
).add_to(m)

title_html = f'''
    <div style="position: absolute; top: 10px; left: 50%; transform: translate(-50%, 0);
                background-color: white; padding: 10px; font-size: 20px; font-weight: bold;
                z-index: 1000; border-radius: 5px; box-shadow: 2px 2px 5px rgba(0,0,0,0.3);">
         Top 5 High Schools in San Francisco
    </div>
    '''
m.get_root().html.add_child(folium.Element(title_html))

m.save('tophs.html')

m
```
